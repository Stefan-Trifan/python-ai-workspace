```python
from sklearn.preprocessing import PolynomialFeatures
import pandas as pd
import numpy as np

df = pd.DataFrame({"x": [np.arange(1,6)]})
print(df)

# transformador polinómico de grado 2
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(df[["x"]])

X_poly # columnas: x, x^2
```
