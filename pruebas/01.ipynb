```python
import numpy as np

x0 = np.array([1,1,1])
x1 = np.array([2,4,6])
x2 = np.array([6,7,8])
y  = np.array([5,7,9])

# Construcción de la matriz de diseño (X y columna de 1s para el intercepto)
X = np.column_stack((x0, x1, x2))

# Ecuación normal
betas = np.linalg.inv(X.T @ X) @ X.T @ y
print(betas)
```
