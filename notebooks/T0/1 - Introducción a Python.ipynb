## Indentaciones en Python

```python
f = 1

while(f < 10):
    print(f * 2)
    f = f + 1

print("Valor Final:", f)
```

## Tipado Dinámico

```python
a = 10      # integer
b = 3.5     # float
c = "texto" # string
d = True    # Boolean
e = None    # null

vars = [a,b,c,d,e]

for var in vars:
    print(type(var))
```

## Funciones Básicas

```python
def suma(a, b):
    """Devuelve la suma de dos números"""
    return a + b

print(suma(5, 7))

help(suma)  # help(): sirve para consultar documentación de funciones
```

## Funciones con Múltiples Valores

```python
def operaciones(a, b):
    return a + b, a - b, a * b

suma, resta, producto = operaciones(6, 3)

print(suma, resta, producto)
```

## Funciones Lambda

```python
cuadrado = lambda x: x**2

print(cuadrado(5))
```

## Funciones Incorporadas en Python

## Operadores Lógicos

```python
a, b = True, False
print(a and b)   # False  (AND lógico)
print(a or b)    # True   (OR lógico)
print(not a)     # False  (NOT lógico)
```

## Condicionales

```python
x = 7

if x > 10:
    print("Mayor que 10")
elif x == 10:
    print("Igual a 10")
else:
    print("Menor que 10")
```

## Operadores con Listas

```python
numeros = [1, 2, 3, 4, 5]

# Acceso e índices
print(numeros[0])
print(numeros[-1])

# Modificación
numeros.append(6)   # añade al final
numeros.remove(3)   # elimina un valor
print(numeros)

# Slicing
print(numeros[1:3])

a = [10, 20, 30, 40, 50]
print(a[-2:])   # últimos dos elementos
print(a[:-2])   # todo excepto los dos últimos
print(a[::-1])  # lista invertida
```

## Diccionarios

```python
alumno = {
    "nombre": "Lucía",
    "edad": 21,
    "curso": "IA"
}

print(alumno["nombre"])

# Añadir o modificar
alumno["edad"] = 22
alumno["nota"] = 9.5
print(alumno)
```

## Creación de Listas con Lógica

```python
# Lista de cuadrados
cuadrados = [x**2 for x in range(5)]
print(cuadrados)

# Filtrar pares
pares = [x for x in range(10) if x % 2 == 0]
print(pares)
```
