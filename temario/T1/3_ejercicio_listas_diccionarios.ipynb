## Datos iniciales

Ejecuta la siguiente celda para cargar los datos.

````python
jugadores = [
    {
        "nombre": "Ana",
        "equipo": "Madrid",
        "partidos": [
            {"rival": "Barcelona", "goles": 2, "asistencias": 1},
            {"rival": "Valencia", "goles": 0, "asistencias": 2},
            {"rival": "Sevilla", "goles": 1, "asistencias": 0}
        ]
    },
    {
        "nombre": "Lucia",
        "equipo": "Barcelona",
        "partidos": [
            {"rival": "Madrid", "goles": 1, "asistencias": 0},
            {"rival": "Valencia", "goles": 3, "asistencias": 1},
            {"rival": "Sevilla", "goles": 0, "asistencias": 1}
        ]
    },
    {
        "nombre": "Marta",
        "equipo": "Madrid",
        "partidos": [
            {"rival": "Barcelona", "goles": 0, "asistencias": 0},
            {"rival": "Valencia", "goles": 2, "asistencias": 2},
            {"rival": "Sevilla", "goles": 2, "asistencias": 1}
        ]
    },
    {
        "nombre": "Clara",
        "equipo": "Sevilla",
        "partidos": [
            {"rival": "Madrid", "goles": 1, "asistencias": 2},
            {"rival": "Barcelona", "goles": 1, "asistencias": 1}
        ]
    }
]

jugadores
````

## 1. Estadísticas individuales

Crea un nuevo diccionario llamado `estadisticas` con esta estructura:

```python
{
    "Ana": {
        "goles": 3,
        "asistencias": 3,
        "puntos": 9
    },
    ...
}
```

Los puntos se calculan así:

- Cada gol = 2 puntos
- Cada asistencia = 1 punto

Por ejemplo, si una jugadora tiene 3 goles y 3 asistencias:

`3 * 2 + 3 = 9 puntos`

````python
# Escribe aquí tu solución

estadisticas = {
    "Ana": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Lucia": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Marta": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Clara": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
}

for jugadora in jugadores:
    for partido in jugadora["partidos"]:
        estadisticas[jugadora["nombre"]]["goles"] += partido["goles"]
        estadisticas[jugadora["nombre"]]["asistencias"] += partido["asistencias"]

for jugadora in estadisticas.values():
    jugadora["puntos"] += jugadora["goles"] * 2 + jugadora["asistencias"]

estadisticas
````

## 2. Encontrar a la MVP

Encuentra automáticamente a la jugadora con más puntos.

El programa debe mostrar algo como:

```text
MVP: Lucia con 12 puntos
```

No escribas manualmente el nombre de la jugadora.

````python
# Escribe aquí tu solución

nombre_ganadora = 0
n_puntos_ganadora = 0

for nombre, datos in estadisticas.items():
    if datos["puntos"] > n_puntos_ganadora:
        n_puntos_ganadora = datos["puntos"]
        nombre_ganadora = nombre

print("MVP: " + nombre_ganadora + " con " + str(n_puntos_ganadora) + " puntos")
print("__________")



# todo con max
nombre_ganadora2 = "pepa"
n_puntos_ganadora2 = 0
"""
{
'Ana':   {'goles': 3, 'asistencias': 3, 'puntos': 9 },
'Lucia': {'goles': 4, 'asistencias': 2, 'puntos': 10},
'Marta': {'goles': 4, 'asistencias': 3, 'puntos': 11 },
'Clara': {'goles': 2, 'asistencias': 3, 'puntos': 7} }
"""

nombre_ganadora2 = max(estadisticas,key=lambda nombre: estadisticas[nombre]["puntos"])
n_puntos_ganadora2 = estadisticas[nombre_ganadora]["puntos"]

print("MVP: " + nombre_ganadora2 + " con " + str(n_puntos_ganadora2) + " puntos")
````

## 3. Agrupar jugadoras por equipo

Crea un diccionario que agrupe a las jugadoras según su equipo.

El resultado debería tener esta estructura:

```python
{
    "Madrid": ["Ana", "Marta"],
    "Barcelona": ["Lucia"],
    "Sevilla": ["Clara"]
}
```

**No puedes escribir manualmente los nombres de los equipos.**

````python
# Escribe aquí tu solución

equipos = {
    "Madrid" : [],
    "Barcelona" : [],
    "Sevilla" : []
}

for jugadora in jugadores:
    match jugadora["equipo"]:
        case "Madrid":
            equipos["Madrid"].append(jugadora["nombre"])
        case "Barcelona":
            equipos["Barcelona"].append(jugadora["nombre"])
        case "Sevilla":
            equipos["Sevilla"].append(jugadora["nombre"])

equipos
````

## 4. Goles totales por equipo

Calcula los goles totales marcados por las jugadoras de cada equipo.

El resultado tendrá esta forma:

```python
{
    "Madrid": 7,
    "Barcelona": 4,
    "Sevilla": 2
}
```

Los valores deben calcularse automáticamente a partir de los datos.

````python
# Escribe aquí tu solución

goles_ciudades = {
    "Madrid": 0,
    "Barcelona": 0,
    "Sevilla": 0
}

for jugadora in jugadores:
    match jugadora["equipo"]:
        case "Madrid":
            for partido in jugadora["partidos"]:
                goles_ciudades["Madrid"] += partido["goles"]
        case "Barcelona":
            for partido in jugadora["partidos"]:
                goles_ciudades["Barcelona"] += partido["goles"]
        case "Sevilla":
            for partido in jugadora["partidos"]:
                goles_ciudades["Sevilla"] += partido["goles"]

goles_ciudades
````

## 5. Ranking de jugadoras

Crea un ranking de jugadoras ordenado de mayor a menor según sus puntos.

Debe tener una estructura similar a:

```python
[
    ("Lucia", 12),
    ("Marta", 11),
    ("Ana", 9),
    ("Clara", 7)
]
```

Los números del ejemplo son orientativos: tu programa debe calcularlos.

````python
# Escribe aquí tu solución

estadisticas = {
    "Ana": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Lucia": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Marta": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
    "Clara": {
        "goles": 0,
        "asistencias": 0,
        "puntos": 0
    },
}

for jugadora in jugadores:
    for partido in jugadora["partidos"]:
        estadisticas[jugadora["nombre"]]["goles"] += partido["goles"]
        estadisticas[jugadora["nombre"]]["asistencias"] += partido["asistencias"]

for jugadora in estadisticas.values():
    jugadora["puntos"] += jugadora["goles"] * 2 + jugadora["asistencias"]

estadisticas

# Creamos un array con tuplas, tipo (nombre, num_goles)

ordenadas = []

for nombre, datos in estadisticas.items():
    nueva = (nombre, datos["puntos"])
    ordenadas.append(nueva)

ordenadas

# Ordenamos el array de tupas
sorted(ordenadas, key=lambda t: t[1], reverse=True)
````

## 6. Rival que más goles ha recibido

Encuentra qué rival ha recibido más goles en total.

Primero tendrás que construir internamente un diccionario parecido a:

```python
{
    "Barcelona": 3,
    "Valencia": 7,
    "Sevilla": 3,
    "Madrid": 2
}
```

Después muestra un mensaje como:

```text
El rival que más goles ha recibido es Valencia con 7 goles
```

````python
# Escribe aquí tu solución

ciudad_goles = {
    "Barcelona": 0,
    "Valencia": 0,
    "Sevilla": 0,
    "Madrid": 0,
}

for jugadora in jugadores:
    for partido in jugadora["partidos"]:
        match partido["rival"]:
            case "Madrid":
                ciudad_goles["Madrid"] += partido["goles"]
            case "Barcelona":
                ciudad_goles["Barcelona"] += partido["goles"]
            case "Sevilla":
                ciudad_goles["Sevilla"] += partido["goles"]
            case "Valencia":
                ciudad_goles["Valencia"] += partido["goles"]

ciudad_goles

ganadora = max(ciudad_goles, key=ciudad_goles.get)
ganadora

print(f"El rival que más goles ha recibido es {ganadora} con {ciudad_goles[ganadora]} goles")
````
